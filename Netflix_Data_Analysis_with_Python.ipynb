```python
# in which Year number of TV Shows and movies released?
```

```python
df["Date_n"] = pd.to_datetime(df["Release_Date"].str.strip())
```

```python
df.head(2)
```

```python
yearly_Count=df['Date_n'].dt.year.value_counts()
yearly_Count
```

```python
yearly_Count.plot(kind="bar", color="skyblue")

plt.bar_label(plt.gca().containers[0])

plt.title("Yearly Count")
plt.xlabel("Year")
plt.ylabel("Count")
plt.tight_layout()
plt.show()
```

```python
# How many TV Shows and Movies are in this dataset ? Show in Bar Graph
```

```python
df.head(2)
```

```python
total_TV_Movies=df.groupby("Category").Category.count()
total_TV_Movies
```

```python
total_TV_Movies.plot(kind="bar", color="skyblue")

plt.bar_label(plt.gca().containers[0])

plt.title("TV_Shows Vs Movies")
plt.xlabel("Category")
plt.ylabel("Count")
plt.tight_layout()
plt.show()
```
